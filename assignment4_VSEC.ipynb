```python
import numpy as np
#np.random.rand (uniform distribution)
x=np.random.rand(4)
print("random numbers:")
print(x)
y=np.random.rand(2,2) #(rows,columns)
print(y)
```

```python
#np.random.randn() (standard normal distribution)
a = np.random.randn(5)
print("Random numbers:")
print(a)
p=np.random.randn(2,3)
print(p)
```

```python
#np.random.randint() (random integers)
a = np.random.randint(1,10,5)  #(low,hign,size) 
#high is not included
print("Random integers:",a)
b = np.random.randint(1,100,(2, 3))
print(b)
```

```python
#np.random.normal()
#np.random.normal(mean, standard_deviation, size)
v=np.random.normal(50,10,5)
print(v)
h=np.random.normal(50,10,(2,3))
print(h)
```

```python
#computing statistics
s=np.random.normal(30,40,10)
print("data:",s)
print("mean:",np.mean(s))
print("median:",np.median(s))
print("standard devision:",np.std(s))
print("minimum:",np.min(s))
print("maximum",np.max(s))
```
