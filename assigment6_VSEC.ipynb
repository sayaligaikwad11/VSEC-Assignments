```python
import pandas as pd
data={
     "Name" : ["Rahul","Priya","Amit","Sneha","Neha"],
    "Department": ["CS","IT","CS","IT","CS"],
    "Marks": [85,None,76,90,None],
    "Age": [19,20,None,21,20]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
```

```python
df["Marks"]=df["Marks"].fillna(0)
print(df)
```

```python
df=df.dropna()
print(df)
```

```python
#filtering
result=df[df["Marks"]>=90]
print(result)
```

```python
result=df[df["Department"]=="CS"]
print(result)
```

```python
result=df[(df["Marks"]>=85) & (df["Department"]=="CS")]
print(result)
```

```python
print(df["Name"])
```

```python
print(df[["Name","Marks"]])
```

```python
df=df.sort_values("Marks",ascending=False)
print(df)
```

```python
data={
     "Name" : ["Rahul","Priya","Amit","Sneha","Neha"],
    "Department": ["CS","IT","CS","IT","CS"],
    "Marks": [85,88,76,90,92],
    "Age": [19,20,18,21,20]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.groupby("Department")["Marks"].sum())
```

```python
print(df.groupby("Department")["Marks"].max())
```

```python
group=df.groupby("Department")
print(group["Marks"].mean())
```

```python
print(df.groupby("Department")["Age"].min())
```
