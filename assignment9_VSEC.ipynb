```python
#assignment9
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as pl
data = {
    "month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jan", "Feb", "Mar", "Apr", "May", "Jun"],   
    "sales": [22, 25, 24, 28, 31, 30,
              18, 20, 23, 21, 26, 27],
    "region": ["North", "North", "North", "North", "North", "North",
               "South", "South", "South", "South", "South", "South"]
}
df = pd.DataFrame(data)
print(df)
```

```python
#sns.lineplot()
import seaborn as sns
import matplotlib.pyplot as plt
sns.lineplot(
    data=df,
    x="month",
    y="sales",
    hue="region"
)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales by Region")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Data for scatter plot
data = {
    "hours_studied": [1, 2, 3, 4, 5, 2, 3, 4, 5, 6],
    "score": [42, 55, 60, 74, 88, 50, 65, 70, 82, 90],
    "gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "attendance": [70, 75, 80, 85, 95, 72, 82, 88, 92, 98]
}
df_student = pd.DataFrame(data)
print(df_student)
```

```python
sns.scatterplot(
    data=df_student,
    x="hours_studied",
    y="score",
    hue="gender",
    size="attendance",
    style="gender"
)
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Data for histogram
data = {
    "score": [
        52, 55, 58,
        61, 63, 65, 67, 68, 69,
        71, 79, 81, 83, 84, 85, 86, 87, 88, 89,
        91, 93, 95
    ]
}
df_scores = pd.DataFrame(data)
print(df_scores)
```

```python
sns.histplot(
    data=df_scores,
    x="score",
    bins=5,
    kde=True,
    color="orange"
)
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Data for box plot
data = {
    "dept": [
        "CS", "CS", "CS",
        "ECE", "ECE", "ECE",
        "Mech", "Mech", "Mech",
        "Civil", "Civil", "Civil"
    ],
    "score": [
        82, 85, 86,
        75, 78, 80,
        70, 72, 74,
        78, 80, 81
    ]
}
df_dept = pd.DataFrame(data)
print(df_dept)
```

```python
sns.boxplot(
    data=df_dept,
    x="dept",
    y="score"
)
plt.xlabel("Department")
plt.ylabel("Score")
plt.title("Score Spread by Department")
plt.show()
```
