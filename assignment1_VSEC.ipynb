```python
import math
print("===python data types===\n")

#integer
a=10
b=20
print("value of a:",a)
print("Value of b:", b)
print("addition:",a+b)
print("subtraction:",a-b)
print("multiplication:",a*b)
print("division:",a/b)
print("floor division:",a//b)
print("modulus:",a%b)
print("power:",a**b)
```

```python
#float
a=24.4
b=22.8
print("value of a:",a)
print("Value of b:", b)
print("addition:",a+b)
print("subtraction:",a-b)
print("multiplication:",a*b)
print("division:",a/b)
print("floor division:",a//b)
print("modulus:",a%b)
print("power:",a**b)
```

```python
#string
s="Nutan maharashtra institute of engineering and technology"
print("upper:",s.upper())
print("lower",s.lower())
print("length:",len(s))
print("replace:",s.replace("and","&"))
print("slice:",s[0:6])
```

```python
#List
list=[10,20,30,40,50,60,70,80,90]
print("original list:",list)
list.append(50)
print("after append:",list)
list.remove(90)
print("after remove:",list)
print("Maximum:",max(list))
print("minimum:",min(list))
print("sum:",sum(list))
```

```python
#tuple
tup=(1,2,3,4,5,6,7,8,9)
print("length:",len(tup))
print("Length:",len(tup))
print("Maximum:",max(tup))
print("Minimum:",min(tup))
```

```python
#dictionary
student = {"Name": "Shivtej", "Age":16, "Course":"Python"}
print("Dictionary:", student)
print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
```

```python
#set
st = {1, 2, 3, 4}
st.add(5)
print("After Add:", st)
st.remove(2)
print("After Remove:", st)
```

```python
#frozen set
fs = frozenset([10, 20, 30, 40])
print("Frozen Set:", fs)
print("Length:", len(fs))
```

```python
#boolean
flag = True
print("Value:", flag)
print("AND Operation:", flag and False)
print("OR Operation:", flag or False)
print("NOT Operation:", not flag)
```

```python
#none type
n=None
print("Value:", n)
print("Type:", type(n))
```

```python
#math module funtion
import math
num=25
print("Square Root:", math.sqrt(num))
print("Power:", math.pow(2, 5))
print("Factorial:", math.factorial(5))
print("Ceiling:", math.ceil(10.2))
print("Floor:", math.floor(10.8))
```
