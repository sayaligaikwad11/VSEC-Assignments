```python
#assignment8
import matplotlib.pyplot as plt
#line plot
x = [1,2,3,4,5]
y = [10,20,15,25,30]

plt.plot(x, y,label="Marks")
plt.title("Marks Progress")
plt.xlabel("Test")
plt.ylabel("Marks")
plt.legend()
plt.grid()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [10,20,15,25,30]
plt.plot(
    x,
    y,
    linestyle="--",
    marker="*",
    linewidth=2,
    label="Sales"
)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid()
plt.show()
```

```python
#sns.linepot()
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = {
    "Student": ["A","B","C","D","E"],
    "Marks": [75, 85, 65, 90, 80]
}
df = pd.DataFrame(data)
sns.lineplot(data=df, x="Student", y="Marks", marker="o")
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
sns.barplot()
sns.barplot(data=df, x="Student", y="Marks")

plt.title("Student Marks Comparison")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.show()
```

```python
sns.histplot()
sns.histplot(data=df, x="Marks", bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```
