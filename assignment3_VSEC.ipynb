```python
#vectorized operations
import numpy as np
s=np.array([2,3,4,5])
p=np.array([11,22,33,55])
#addition
result=np.add(s,p)
print(result)
#np.sqrt
r=np.array([1,4,9,16,25])
print("sqrt=",np.sqrt(r))
#np.exp()
u=np.array([2,3,4,6,7])
print(np.exp(u))

#broadcasting
t=np.array([5,6,7,3,4,23])
print(t+5)
h=np.array([6,7,8,9,10])
v=np.array([[2,3,4,5,6],
      [4,5,6,4,6]])
print(h+v)

i = np.array([[1],
              [2],
              [3]])
m = np.array([10, 20, 30])
print(i + m)
```
